### 0. Importer pandas

**Déf :** `Pandas est une bibliothèque open-source du langage Python conçue pour la manipulation et l'analyse de données. Elle permet de travailler facilement avec des données tabulaires (lignes et colonnes), un peu comme un tableau Excel ou une table SQL`

```python
import pandas as pd
```

### 1.1 Création de la Series

La fonction `pd.Series()` permet de créer une Series à partir d'une liste de valeurs.

Chaque élément de la liste devient une valeur de la series et reçoit automatiquement un index.

```python
temperatures = pd.Series([24.5, 26.1, 25.8, 27.3])

temperatures
```

### 1.2 Afficher l'index de la Series

L'index permet d'identifier chaque valeur d'une Series.

Par défaut, Pandas crée automatiquement un index numérique commençant à `0`.

```python
temperatures.index
```

### 1.3 Afficher la première valeur

La méthode `iloc[]` permet d'accéder à un élément selon sa position.

La première valeur d'une Series se trouve à la position `0`.

```python
temperatures.iloc[0]
```

### 1.4 Afficher la dernière valeur

Pour accéder au dernier élément d'une Series, on utilise l'indice `-1`.

Cette méthode fonctionne quelle que soit la taille de la Series.

```python
temperatures.iloc[-1]
```

### 1.5 Créer une series avec un index personnalisé (méthode 1)

Il est possible de définir un index personnalisé lors de la création d'une series.

Chaque température est ici associée à son heure de mesure.

```python
temperatures_horaires = pd.Series(
    data=[24.5, 26.1, 25.8, 27.3],
    index=["12h", "13h", "14h", "15h"]
)

temperatures_horaires
```

### 1.5 Créer une series avec un index personnalisé (méthode 2)

Une autre approche consiste à créer la series, puis à modifier son index après sa création.

Chose qui produit le même résultat que la précédente.

```python
temperatures_horaires_2 = pd.Series([24.5, 26.1, 25.8, 27.3])

temperatures_horaires_2.index = ["12h", "13h", "14h", "15h"]

temperatures_horaires_2
```

### 1.6 Afficher la première et la dernière valeur

Même avec un index personnalisé, `iloc[]` permet d'accéder aux valeurs selon leur position.

La première valeur est à la position `0` et la dernière à la position `-1`.

```python
print("Première valeur :", temperatures_horaires.iloc[0])
print("Dernière valeur :", temperatures_horaires.iloc[-1])
```

### 5.1 Ajouter la colonne `temperature_fahrenheit`

Il est possible de créer une nouvelle colonne à partir d'une colonne existante.

La conversion de la température de **Celsius** vers **Fahrenheit** s'effectue avec la formule :

**F = (C × 9 / 5) + 32**

```python
df["temperature_fahrenheit"] = (df["temperature"] * 9 / 5) + 32

df.head()
```

### 5.2 Ajouter la colonne `niveau_temperature`

La fonction `np.where()` permet de créer une nouvelle colonne en fonction d'une condition.

Une température supérieure à **30°C** sera considérée comme **"Élevée"**, sinon elle sera considérée comme **"Normale"**.

```python
import numpy as np

df["niveau_temperature"] = np.where(
    df["temperature"] > 30,
    "Élevée",
    "Normale"
)

df.head()
```

### 5.3 Renommer la colonne `humidite`

La méthode `rename()` permet de modifier le nom d'une ou de plusieurs colonnes.

Ici, la colonne **humidite** est renommée en **humidite_relative**.

```python
df.rename(
    columns={"humidite": "humidite_relative"},
    inplace=True
)

df.head()
```

### 5.4 Supprimer la colonne `temperature_fahrenheit`

La méthode `drop()` permet de supprimer une ou plusieurs colonnes d'un DataFrame.

La colonne créée précédemment va etre retirées.

```python
df.drop(
    columns="temperature_fahrenheit",
    inplace=True
)

df.head()
```

### 6.2 Récupérer les lignes où la température est supérieure à 30°C et l'humidité supérieure à 70%

### 9.1 Afficher le nombre de valeurs manquantes par colonne

La méthode `isna()` permet de détecter les valeurs manquantes.

```python
df.isna().sum()
```

### 9.2 Afficher le nombre total de valeurs manquantes

Après avoir compté les valeurs manquantes par colonne, il est possible d'obtenir le total en additionnant ces résultats.

```python
df.isna().sum().sum()
```

### 9.3 Afficher le taux de valeurs manquantes

Le taux de valeurs manquantes permet de mesurer la proportion de données absentes dans chaque colonne.

```python
(df.isna().mean() * 100).round(2)
```

### 9.4 Afficher les lignes contenant des valeurs manquantes

Le filtrage permet de retrouver uniquement les lignes comportant au moins une valeur manquante.

```python
df[df.isna().any(axis=1)]
```

### 9.5 Remplacer les valeurs manquantes de la température et de l'humidité

Les valeurs manquantes des colonnes **temperature** et **humidite_relative** sont remplacées par la moyenne de leur colonne.

Cette méthode permet de conserver les lignes tout en limitant l'impact des données manquantes.

```python
df["temperature"] = df["temperature"].fillna(df["temperature"].mean())

df["humidite_relative"] = df["humidite_relative"].fillna(
    df["humidite_relative"].mean()
)

df[["temperature", "humidite_relative"]].isna().sum()
```

### 9.6 Remplacer les valeurs manquantes de la consommation

La médiane est moins sensible aux valeurs extrêmes que la moyenne.

Elle est utilisée ici pour remplacer les valeurs manquantes de la colonne **consommation**.

```python
df["consommation"] = df["consommation"].fillna(
    df["consommation"].median()
)

df["consommation"].isna().sum()
```

### 9.7 Remplacer les valeurs manquantes de l'état

Les valeurs manquantes de la colonne **etat** sont remplacées par la chaîne de caractères **"INCONNU"**.

```python
df["etat"] = df["etat"].fillna("INCONNU")

df["etat"].isna().sum()
```
