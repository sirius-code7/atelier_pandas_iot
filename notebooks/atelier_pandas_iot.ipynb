### 0. Importer pandas

**Déf :** `Pandas est une bibliothèque open-source du langage Python conçue pour la manipulation et l'analyse de données. Elle permet de travailler facilement avec des données tabulaires (lignes et colonnes), un peu comme un tableau Excel ou une table SQL`

```python
import pandas as pd
```

### 1.1 Création de la Series

La fonction `pd.Series()` permet de créer une Series à partir d'une liste de valeurs.

Chaque élément de la liste devient une valeur de la series et reçoit automatiquement un index.

```python
temperatures = pd.Series([24.5, 26.1, 25.8, 27.3])

temperatures
```

### 1.2 Afficher l'index de la Series

L'index permet d'identifier chaque valeur d'une Series.

Par défaut, Pandas crée automatiquement un index numérique commençant à `0`.

```python
temperatures.index
```

### 1.3 Afficher la première valeur

La méthode `iloc[]` permet d'accéder à un élément selon sa position.

La première valeur d'une Series se trouve à la position `0`.

```python
temperatures.iloc[0]
```

### 1.4 Afficher la dernière valeur

Pour accéder au dernier élément d'une Series, on utilise l'indice `-1`.

Cette méthode fonctionne quelle que soit la taille de la Series.

```python
temperatures.iloc[-1]
```

### 1.5 Créer une series avec un index personnalisé (méthode 1)

Il est possible de définir un index personnalisé lors de la création d'une series.

Chaque température est ici associée à son heure de mesure.

```python
temperatures_horaires = pd.Series(
    data=[24.5, 26.1, 25.8, 27.3],
    index=["12h", "13h", "14h", "15h"]
)

temperatures_horaires
```

### 1.5 Créer une series avec un index personnalisé (méthode 2)

Une autre approche consiste à créer la series, puis à modifier son index après sa création.

Chose qui produit le même résultat que la précédente.

```python
temperatures_horaires_2 = pd.Series([24.5, 26.1, 25.8, 27.3])

temperatures_horaires_2.index = ["12h", "13h", "14h", "15h"]

temperatures_horaires_2
```

### 1.6 Afficher la première et la dernière valeur

Même avec un index personnalisé, `iloc[]` permet d'accéder aux valeurs selon leur position.

La première valeur est à la position `0` et la dernière à la position `-1`.

```python
print("Première valeur :", temperatures_horaires.iloc[0])
print("Dernière valeur :", temperatures_horaires.iloc[-1])
```

### 2.1 Créer et afficher un DataFrame

Un **DataFrame** est une structure de données bidimensionnelle composée de lignes et de colonnes.

Il peut être créé à partir d'un dictionnaire où chaque clé représente une colonne et chaque valeur une liste de données.

```python
data = {
    "capteur": ["C001", "C002", "C003"],
    "temperature": [24.5, 25.1, 26.3],
    "humidite": [65, 67, 70]
}

df_test = pd.DataFrame(data)

df_test
```
